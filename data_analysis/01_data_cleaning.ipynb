## Check Missing Value

```python
df.isnull().sum()
```

### Check Age Column

```python
# Check data where 'Age' is null/NaN
df[df['Age'].isnull()]
```

```python
# Check value in 'Age'
df['Age'].value_counts()
```

```python
# Clean ' yrs' and convert to float temporary
df['Age'] = df['Age'].str.extract(r'(\d+)').astype(float)

df['Age'].value_counts()
```

```python
# mean, median, modus in 'Age'
print(df['Age'].mean())
print(df['Age'].median())
print(df['Age'].mode()[0])
```

```python
# Fill the NaN 'Age' with median value of 'Age'
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Age'] = df['Age'].astype(int)

df['Age'].isnull().sum()
```

```python
df['Age'].value_counts()
```

```python
df.info()
```

## Cleaning Bagian 1

### Check Gender Column

```python
df[df['Gender'].isnull()]
```

```python
df['Gender'].isnull().sum()
```

```python
df['Gender'].value_counts()
```

```python
# Gender is only two (male or female).
# But sometimes, there is a option to not display a gender. So, in addition to male and female, i will replace with 'Not Display'. (I don't support LGBTQ)
df['Gender'] = df['Gender'].replace({
   'male' : 'Male', 'M': 'Male', 'MALE': 'Male',
   'F': 'Female', 'FEMALE': 'Female', 'female': 'Female',
   'Other': 'Not Display', 'other': 'Not Display', 'Non-binary': 'Not Display'
})

df['Gender'] = df['Gender'].fillna('Not Display')

df['Gender'].value_counts()
```

### Check Country Column

```python
df[df['Country'].isnull()]
```

```python
df['Country'].isnull().sum()
```

```python
df['Country'].value_counts()
```

```python
# I will fill the NaN value in column 'Country' with 'Not Display', make it consist with other column like 'Gender'.
# And replace the lowercase with capital in the first alphabet

df['Country'] = df['Country'].fillna('Not Display')
df['Country'] = df['Country'].replace({
   'brazil': 'Brazil', 'india': 'India', 'canada': 'Canada',
   'france': 'France', 'U.K.': 'United Kingdom',
   'United States of America': 'United States',
   'philippines': 'Philippines', 'australia': 'Australia'
})

df['Country'].value_counts()
```

```python
df['Country'].isnull().sum()
```

### Checking Monthly_Income

```python
df['Monthly_Income'].tail(60)
```

```python
# Extranct digits, change data type to float, fill missing value with median, and change data type to integer
df['Monthly_Income'] = df['Monthly_Income'].str.extract(r'(\d+)').astype(float)
df['Monthly_Income'] = df['Monthly_Income'].fillna(df['Monthly_Income'].median())
df['Monthly_Income'] = df['Monthly_Income'].astype(int)
```
