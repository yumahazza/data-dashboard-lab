```python
import matplotlib.pyplot as plt
import pandas as pd

# Data Dummy
data = {
    'Bulan': ['Jan', 'Feb', 'Mar', 'Apr', 'Mei'],
    'Penjualan': [120, 150, 280, 190, 250],
    'Target': [100, 100, 150, 150, 200]
}
df = pd.DataFrame(data)

# Bikin canvas dengan 1 baris & 2 kolom (Subplots)
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# --- CHART 1: Bar Chart Penjualan ---
colors = ['#e74c3c' if bulan == 'Mar' else '#2b5c8f' for bulan in df['Bulan']]
bars = ax[0].bar(df['Bulan'], df['Penjualan'], color=colors)
ax[0].set_title('Total Penjualan per Bulan', fontsize=12, fontweight='bold')
ax[0].set_xlabel('Bulan')
ax[0].set_ylabel('Penjualan (Juta Rp)')
ax[0].grid(axis='y', linestyle='--', alpha=0.5)

# Annotation: Sorot angka tertinggi (Maret = 280)
ax[0].annotate('Puncak tertinggi!', 
               xy=(2, 280), xytext=(1.5, 300),
               arrowprops=dict(facecolor='red', shrink=0.05, width=1, headwidth=6))

# --- CHART 2: Line Chart vs Target ---
ax[1].plot(df['Bulan'], df['Penjualan'], marker='o', color='#2b5c8f', label='Realisasi')
ax[1].plot(df['Bulan'], df['Target'], marker='s', color='#e74c3c', linestyle='--', label='Target')
ax[1].set_title('Realisasi vs Target Penjualan', fontsize=12, fontweight='bold')
ax[1].set_xlabel('Bulan')
ax[1].set_ylabel('Juta Rp')
ax[1].legend()
ax[1].grid(True, linestyle=':', alpha=0.6)

# Merapikan jarak antar chart
plt.tight_layout()
plt.show()
```
