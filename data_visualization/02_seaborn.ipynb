## Import Library

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## Load Dataset

```python
df = pd.read_csv('../data_visualization/sales_data.csv')

df.info()
df.describe()
```

```python
df.isnull().sum()
```

## Box Plot & Heatmap

```python
sns.set_theme(style='whitegrid')

fig, ax = plt.subplots(1, 2, figsize=(14, 5))

kolom_kategori = 'Product_Category' 
kolom_numerik = 'Unit_Price' 

# Box Plot
sns.boxplot(
    data=df,
    x=kolom_kategori,
    y=kolom_numerik,
    palette='Set2',
    ax=ax[0],
)
ax[0].set_title(
    f'Sebaran {kolom_numerik} Berdasarkan {kolom_kategori}',
    fontsize=12,
    fontweight='bold',
)

numeric_df = df.select_dtypes(include=['number'])
corr_matrix = numeric_df.corr()

# Heatmap
sns.heatmap(corr_matrix, annot=True, cmap='Blues', fmt='.2f', linewidths=0.5, ax=ax[1])
ax[1].set_title(
    'Heatmap Korelasi Variabel Numerik', fontsize=12, fontweight='bold'
)

plt.tight_layout()
plt.show()
```
