## Import Library

```python
import pandas as pd
import plotly.express as px
```

## Load Dataset

```python
df = pd.read_csv('../sales_data.csv')

df.info()
df.describe()
```

```python
cat_prod_amount = df.groupby('Product_Category')['Sales_Amount'].sum().reset_index()

graph_bar = px.bar(
   cat_prod_amount,
   x='Product_Category',
   y='Sales_Amount',
   color='Product_Category',
   title='Jumlah Penjualan per Kategori Produk',
   labels={'Sales_Amount': 'Jumlah Penjualan ($)', 'Product_Category': 'Kategori'}
)

graph_bar.show()
```

```python
fig_scatter = px.scatter(
    df,
    x='Unit_Price',
    y='Sales_Amount',
    color='Product_Category',
    hover_data=['Sales_Rep', 'Region'],  # Menampilkan info tambahan saat kursor diarahkan
    title='Hubungan Unit Price vs Sales Amount'
)

fig_scatter.show()
```
