## Import Library

```python
import pandas as pd
import plotly.express as px
```

## Load Dataset

```python
df = pd.read_csv('../data_visualization/sales_data.csv')

df.info()
df.describe()
```

## Interactive Bar Chart

```python
cat_prod_amount = df.groupby('Product_Category')['Sales_Amount'].sum().reset_index()

graph_bar = px.bar(
   cat_prod_amount,
   x='Product_Category',
   y='Sales_Amount',
   color='Product_Category',
   title='Jumlah Penjualan per Kategori Produk',
   labels={'Sales_Amount': 'Jumlah Penjualan ($)', 'Product_Category': 'Kategori'}
)

graph_bar.show()
```

## Interactive Scatter Plot

```python
fig_scatter = px.scatter(
    df,
    x='Unit_Price',
    y='Sales_Amount',
    color='Product_Category',
    hover_data=['Sales_Rep', 'Region'],  # Menampilkan info tambahan saat kursor diarahkan
    title='Hubungan Unit Price vs Sales Amount'
)

fig_scatter.show()
```

## Tree Map

```python
fig_tree = px.treemap(
    df,
    path=['Region', 'Product_Category', 'Sales_Rep'],
    values='Sales_Amount',
    color='Sales_Amount',
    color_continuous_scale='Viridis',
    title='Breakdown Penjualan: Region > Kategori > Sales Rep'
)

fig_tree.update_layout(
    margin=dict(t=50, l=10, r=10, b=10),
    font=dict(family="Arial", size=14)
)

fig_tree.show()
```

## Animated Chart

```python
# Melihat Perubahan Data Antar Periode
# Ekstrak Bulan & Tahun dari Sale_Date
df['Year_Month'] = pd.to_datetime(df['Sale_Date']).dt.to_period('M').astype(str)

fig_anim = px.scatter(
    df.sort_values('Year_Month'),
    x='Unit_Price',
    y='Sales_Amount',
    animation_frame='Year_Month',
    color='Product_Category',
    hover_name='Sales_Rep',
    size='Quantity_Sold',
    range_x=[0, 5000],
    range_y=[0, 10000],
    title='Dinamika Penjualan Bulanan'
)

fig_anim.show()
```
